```python
import pandas as pd
df = pd.read_csv(r'C:\Users\nitha\Documents\NITHA\STUDY\ICT PROJECT\ProjectSmartCareerAI\data\Resume.csv')
df
```

```python
df['Job_Title'] = (
    df['Resume_str']
    .str.split(r'(?i)summary')
    .str[0]
    .str.strip()
)
```

```python
df['Job_Title'] = df['Job_Title'].str.split('\n').str[0].str.strip()
```

```python
df['Job_Title'] = (
    df['Job_Title']
    .str.split(r'\s{2,}', regex=True)
    .str[0]
    .str.strip()
)
```

```python
remove_titles = [
    '', 'Y', 'R', 'S', 'Objective', 'AS', 'EA', 'WORKING',
    'RANK: SGT',

    'Kimberly', 'TW', 'Marilyn', 'Kpandipou', 'G',
    'Rachael', 'Rachel', 'I', 'IMPACT', 'ADVISOR',
    'OWNER', 'WS', 'AM', 'O', 'Z', 'JS', 'GM',
    'Camryn', 'Pavithra', 'RM', 'GC',

    'RN', 'HEAD', 'Zachory', 'KICHEN',

    'GENERAL', 'COORDINATOR', 'Qualifications',
    'PRESIDENT', 'SR.', 'HEALTHCARE', 'BUSINESS',
    'Christopher', 'CHEF',

    'UT-TSU AGRICULTURE / 4-H EXTENSION AGENT TRAINEE',

    'FELLOW: CREATING LIVELIHOODS VIA AGRICULTURE PROJECT, VOLUNTEERING FOR INTERNATIONAL PROFESSIONALS',

    '25LIVE FUNCTIONAL ADMINISTRATOR/SUPERVISOR OF I.T. HELP DESK',

    'JAMAICA- APICULTURE EXTENSION SPECIALIST'
]

df = df[~df['Job_Title'].isin(remove_titles)]
```

```python
df['Job_Title'] = (
    df['Job_Title']
    .str.split(r',|\s{2,}', regex=True)
    .str[0]
    .str.strip()
)
```

```python
df = df[~df['Job_Title'].str.match(r'^\d', na=False)]
```

```python
df['Job_Title'] = df['Job_Title'].str.split('/').str[0].str.strip()
```

```python
# Invalid/non-job titles to remove
invalid_titles = {
    "RANK: SGT", "GOVERNMENT RELATIONS", "HEAD", "ABOUT",
    "Professional", "Highlights", "Work History", "READING",
    "BANKING", "MARKETING", "MEDIA", "FINANCE", "AUDIO",
    "PERSONAL", "P", "V", "EP", "D", "QA", "CS", "AP",
    "PMC", "RN", "KICHEN", "EIGHTH", "TSO"
}

# Remove rows
df = df[~df['Job_Title'].astype(str).str.strip().isin(invalid_titles)]

# remove very short/non-meaningful titles
df = df[df['Job_Title'].astype(str).str.len() > 3]
```
